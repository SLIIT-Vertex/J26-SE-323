```python
import pandas as pd

file_path = "/kaggle/input/datasets/rasangabawanthi/mindbridge-assistments-2012-2013/2012-2013-data-with-predictions-4-final.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Columns in dataset:")
print(df.columns.tolist())
```

```python
target_cols = [
    'Average_confidence(FRUSTRATED)',
    'Average_confidence(CONFUSED)',
    'Average_confidence(CONCENTRATING)',
    'Average_confidence(BORED)'
]

print("Dataset shape:", df.shape)

print("\nTarget statistics:")
print(df[target_cols].describe())

print("\nMissing target values:")
print(df[target_cols].isnull().sum())

print("\nConfusion values:")
print(df['Average_confidence(CONFUSED)'].value_counts(dropna=False).head(20))
```

```python
candidate_features = [
    'correct',
    'bottom_hint',
    'hint_count',
    'actions',
    'attempt_count',
    'ms_first_response',
    'overlap_time',
    'first_action',
    'position',
    'original'
]

print(df[candidate_features].dtypes)

print("\nMissing values:")
print(df[candidate_features].isnull().sum())

df[candidate_features].head(10)
```

```python
target = 'Average_confidence(CONFUSED)'

print("Total rows:", len(df))
print("Unique learners:", df['user_id'].nunique())

print("\n--- Confusion class balance at different thresholds ---")

thresholds = [0.0, 0.25, 0.50, 0.75]

for t in thresholds:
    if t == 0:
        positive = (df[target] > 0).sum()
    else:
        positive = (df[target] >= t).sum()

    negative = len(df) - positive

    print(f"\nThreshold: {t}")
    print("Confused:", positive)
    print("Not confused:", negative)
    print("Confused %:", round((positive / len(df)) * 100, 2))
```

```python
# ---------------------------------------
# Step 1: Create provisional target label
# ---------------------------------------

target_col = 'Average_confidence(CONFUSED)'

df_model = df[
    [
        'user_id',
        'correct',
        'bottom_hint',
        'hint_count',
        'attempt_count',
        'ms_first_response',
        'overlap_time',
        'first_action',
        'position',
        'original',
        target_col
    ]
].copy()

# High-confidence confusion = 1
# Not confused / lower confidence = 0
df_model['confused_label'] = (df_model[target_col] >= 0.5).astype(int)

# Keep track of missing bottom_hint values
df_model['bottom_hint_missing'] = df_model['bottom_hint'].isna().astype(int)

print("Shape:", df_model.shape)

print("\nClass balance:")
print(df_model['confused_label'].value_counts())

print("\nClass percentage:")
print(df_model['confused_label'].value_counts(normalize=True) * 100)

print("\nMissing values:")
print(df_model.isnull().sum())

df_model.head()
```

```python
from sklearn.model_selection import train_test_split
import numpy as np

# Get unique learners
unique_users = df_model['user_id'].unique()

print("Total unique learners:", len(unique_users))

# 70% Train, 30% Temporary
train_users, temp_users = train_test_split(
    unique_users,
    test_size=0.30,
    random_state=42
)

# Split remaining 30% into 15% Validation + 15% Test
val_users, test_users = train_test_split(
    temp_users,
    test_size=0.50,
    random_state=42
)

print("\nLearner counts")
print("Train learners:", len(train_users))
print("Validation learners:", len(val_users))
print("Test learners:", len(test_users))

# Create row masks
train_mask = df_model['user_id'].isin(train_users)
val_mask = df_model['user_id'].isin(val_users)
test_mask = df_model['user_id'].isin(test_users)

print("\nRow counts")
print("Train rows:", train_mask.sum())
print("Validation rows:", val_mask.sum())
print("Test rows:", test_mask.sum())

print("\nConfusion percentages")
print("Train:", round(df_model.loc[train_mask, 'confused_label'].mean() * 100, 2), "%")
print("Validation:", round(df_model.loc[val_mask, 'confused_label'].mean() * 100, 2), "%")
print("Test:", round(df_model.loc[test_mask, 'confused_label'].mean() * 100, 2), "%")

# Check for learner leakage
print("\nLeakage checks")
print("Train ∩ Validation:", len(set(train_users) & set(val_users)))
print("Train ∩ Test:", len(set(train_users) & set(test_users)))
print("Validation ∩ Test:", len(set(val_users) & set(test_users)))
```

```python
# ============================================
# STEP: Prepare Train / Validation / Test Data
# ============================================

feature_cols = [
    'correct',
    'bottom_hint',
    'hint_count',
    'attempt_count',
    'ms_first_response',
    'overlap_time',
    'first_action',
    'position',
    'original',
    'bottom_hint_missing'
]

target_col = 'confused_label'

# Create separate datasets
train_df = df_model.loc[train_mask, feature_cols + [target_col]].copy()
val_df   = df_model.loc[val_mask, feature_cols + [target_col]].copy()
test_df  = df_model.loc[test_mask, feature_cols + [target_col]].copy()

print("Train shape:", train_df.shape)
print("Validation shape:", val_df.shape)
print("Test shape:", test_df.shape)
```

```python
# ============================================
# Handle missing values WITHOUT data leakage
# ============================================

bottom_hint_median = train_df['bottom_hint'].median()

print("Train median for bottom_hint:", bottom_hint_median)

train_df['bottom_hint'] = train_df['bottom_hint'].fillna(bottom_hint_median)
val_df['bottom_hint']   = val_df['bottom_hint'].fillna(bottom_hint_median)
test_df['bottom_hint']  = test_df['bottom_hint'].fillna(bottom_hint_median)

print("\nMissing values after filling:")
print("Train:", train_df.isnull().sum().sum())
print("Validation:", val_df.isnull().sum().sum())
print("Test:", test_df.isnull().sum().sum())
```

```python
# ============================================
# Create X and y
# ============================================

X_train = train_df[feature_cols]
y_train = train_df[target_col]

X_val = val_df[feature_cols]
y_val = val_df[target_col]

X_test = test_df[feature_cols]
y_test = test_df[target_col]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTraining class percentages:")
print(y_train.value_counts(normalize=True) * 100)
```

```python
# ============================================
# Calculate class imbalance weight
# ============================================

negative_count = (y_train == 0).sum()
positive_count = (y_train == 1).sum()

scale_pos_weight = negative_count / positive_count

print("Negative samples:", negative_count)
print("Positive samples:", positive_count)
print("scale_pos_weight:", round(scale_pos_weight, 2))
```

```python
from xgboost import XGBClassifier

# --------------------------------------------
# Quick training subset for first experiment
# --------------------------------------------

sample_size = 600_000

train_sample = train_df.sample(
    n=sample_size,
    random_state=42
)

X_train_sample = train_sample[feature_cols]
y_train_sample = train_sample[target_col]

print("Training sample shape:", X_train_sample.shape)
print("\nClass distribution:")
print(y_train_sample.value_counts())

print("\nPositive percentage:")
print(y_train_sample.mean() * 100)
```

```python
model = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,

    subsample=0.8,
    colsample_bytree=0.8,

    scale_pos_weight=scale_pos_weight,

    objective='binary:logistic',
    eval_metric='aucpr',

    tree_method='hist',

    random_state=42,
    n_jobs=-1
)

model.fit(
    X_train_sample,
    y_train_sample,
    eval_set=[(X_val, y_val)],
    verbose=25
)
```

```python
import numpy as np
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score,
    f1_score
)

# Probability of confusion
val_prob = model.predict_proba(X_val)[:, 1]

# Initial threshold
threshold = 0.50

val_pred = (val_prob >= threshold).astype(int)

print("ROC-AUC:",
      round(roc_auc_score(y_val, val_prob), 4))

print("PR-AUC:",
      round(average_precision_score(y_val, val_prob), 4))

print("Macro F1:",
      round(f1_score(y_val, val_pred, average='macro'), 4))

print("\nClassification Report:")
print(
    classification_report(
        y_val,
        val_pred,
        digits=4
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, val_pred))
```

```python
# ============================================
# Learner-specific historical baseline dataset
# ============================================

baseline_df = df[
    [
        'user_id',
        'start_time',
        'correct',
        'bottom_hint',
        'hint_count',
        'attempt_count',
        'ms_first_response',
        'overlap_time',
        'first_action',
        'position',
        'original',
        'Average_confidence(CONFUSED)'
    ]
].copy()

# Target
baseline_df['confused_label'] = (
    baseline_df['Average_confidence(CONFUSED)'] >= 0.5
).astype(int)

# Missing-value indicator
baseline_df['bottom_hint_missing'] = (
    baseline_df['bottom_hint'].isna().astype(int)
)

# Sort each learner chronologically
baseline_df = baseline_df.sort_values(
    ['user_id', 'start_time']
).reset_index(drop=True)

print(baseline_df.shape)
baseline_df.head()
```

```python
# ============================================
# Historical learner baselines
# Only PREVIOUS interactions are used
# ============================================

group = baseline_df.groupby('user_id', sort=False)

# Number of previous interactions
baseline_df['previous_interactions'] = group.cumcount()

# Cumulative sums BEFORE current interaction
baseline_df['previous_response_sum'] = (
    group['ms_first_response'].cumsum()
    - baseline_df['ms_first_response']
)

baseline_df['previous_attempt_sum'] = (
    group['attempt_count'].cumsum()
    - baseline_df['attempt_count']
)

baseline_df['previous_hint_sum'] = (
    group['hint_count'].cumsum()
    - baseline_df['hint_count']
)

# Learner-specific historical means
baseline_df['learner_response_baseline'] = (
    baseline_df['previous_response_sum']
    / baseline_df['previous_interactions']
)

baseline_df['learner_attempt_baseline'] = (
    baseline_df['previous_attempt_sum']
    / baseline_df['previous_interactions']
)

baseline_df['learner_hint_baseline'] = (
    baseline_df['previous_hint_sum']
    / baseline_df['previous_interactions']
)

baseline_df[
    [
        'user_id',
        'previous_interactions',
        'ms_first_response',
        'learner_response_baseline',
        'attempt_count',
        'learner_attempt_baseline',
        'hint_count',
        'learner_hint_baseline'
    ]
].head(20)
```

```python
# ============================================
# Deviation from learner's own baseline
# ============================================

baseline_df['response_time_deviation'] = (
    baseline_df['ms_first_response']
    - baseline_df['learner_response_baseline']
)

baseline_df['attempt_deviation'] = (
    baseline_df['attempt_count']
    - baseline_df['learner_attempt_baseline']
)

baseline_df['hint_deviation'] = (
    baseline_df['hint_count']
    - baseline_df['learner_hint_baseline']
)

baseline_df[
    [
        'user_id',
        'ms_first_response',
        'learner_response_baseline',
        'response_time_deviation',
        'attempt_count',
        'learner_attempt_baseline',
        'attempt_deviation'
    ]
].head(20)
```

```python
print(
    baseline_df['previous_interactions']
    .describe()
)

print("\nLearners/rows with at least 5 previous interactions:")

usable = baseline_df['previous_interactions'] >= 5

print("Rows:", usable.sum())
print(
    "Percentage:",
    round(usable.mean() * 100, 2),
    "%"
)
```

```python
import numpy as np

# Log transformed time features
baseline_df['log_response_time'] = np.log1p(
    baseline_df['ms_first_response'].clip(lower=0)
)

baseline_df['log_overlap_time'] = np.log1p(
    baseline_df['overlap_time'].clip(lower=0)
)

# Relative deviations
baseline_df['response_time_ratio'] = (
    baseline_df['ms_first_response'] /
    (baseline_df['learner_response_baseline'] + 1)
)

baseline_df['attempt_ratio'] = (
    baseline_df['attempt_count'] /
    (baseline_df['learner_attempt_baseline'] + 1)
)

baseline_df['hint_ratio'] = (
    baseline_df['hint_count'] /
    (baseline_df['learner_hint_baseline'] + 1)
)

print("New learner-specific features created.")
```

```python
# Require at least 5 previous interactions
personalized_df = baseline_df[
    baseline_df['previous_interactions'] >= 5
].copy()

print("Usable rows:", len(personalized_df))
print(
    "Percentage retained:",
    round(len(personalized_df) / len(baseline_df) * 100, 2),
    "%"
)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

def threshold_search(y_true, probabilities):
    rows = []

    for t in np.arange(0.50, 0.91, 0.02):
        pred = (probabilities >= t).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_true,
            pred
        ).ravel()

        rows.append({
            "threshold": round(t, 2),
            "precision_confused": precision_score(
                y_true, pred, zero_division=0
            ),
            "recall_confused": recall_score(
                y_true, pred, zero_division=0
            ),
            "f1_confused": f1_score(
                y_true, pred, pos_label=1
            ),
            "macro_f1": f1_score(
                y_true, pred, average='macro'
            ),
            "false_positives": fp,
            "false_negatives": fn
        })

    result = pd.DataFrame(rows)

    return result, result.loc[
        result['macro_f1'].idxmax()
    ]


v1_thresholds, v1_best = threshold_search(
    y_val_matched,
    val_prob_v1
)

v2_thresholds, v2_best = threshold_search(
    y_val_v2,
    val_prob_v2
)

print("V1 BEST")
print(v1_best)

print("\nV2 BEST")
print(v2_best)
```

classification performance, although it slightly reduced false positives.
A recent-history learner baseline will therefore be investigated.

```python
# ============================================
# Recent learner baseline - previous 20 events
# ============================================

WINDOW = 20

# Ensure chronological ordering
baseline_df = baseline_df.sort_values(
    ['user_id', 'start_time']
).reset_index(drop=True)

# Log response time first because raw response time has large outliers
baseline_df['log_response_time'] = np.log1p(
    baseline_df['ms_first_response'].clip(lower=0)
)

# Previous-value streams
baseline_df['prev_log_rt'] = (
    baseline_df.groupby('user_id')['log_response_time'].shift(1)
)

baseline_df['prev_attempt'] = (
    baseline_df.groupby('user_id')['attempt_count'].shift(1)
)

baseline_df['prev_hint'] = (
    baseline_df.groupby('user_id')['hint_count'].shift(1)
)

baseline_df['prev_correct'] = (
    baseline_df.groupby('user_id')['correct'].shift(1)
)
```

```python
# ============================================
# Rolling averages from ONLY previous events
# ============================================

baseline_df['recent_rt_baseline'] = (
    baseline_df.groupby('user_id')['prev_log_rt']
    .transform(
        lambda x: x.rolling(
            window=WINDOW,
            min_periods=5
        ).mean()
    )
)

baseline_df['recent_attempt_baseline'] = (
    baseline_df.groupby('user_id')['prev_attempt']
    .transform(
        lambda x: x.rolling(
            window=WINDOW,
            min_periods=5
        ).mean()
    )
)

baseline_df['recent_hint_baseline'] = (
    baseline_df.groupby('user_id')['prev_hint']
    .transform(
        lambda x: x.rolling(
            window=WINDOW,
            min_periods=5
        ).mean()
    )
)

baseline_df['recent_accuracy_baseline'] = (
    baseline_df.groupby('user_id')['prev_correct']
    .transform(
        lambda x: x.rolling(
            window=WINDOW,
            min_periods=5
        ).mean()
    )
)
```

```python
baseline_df['recent_rt_deviation'] = (
    baseline_df['log_response_time']
    - baseline_df['recent_rt_baseline']
)

baseline_df['recent_attempt_deviation'] = (
    baseline_df['attempt_count']
    - baseline_df['recent_attempt_baseline']
)

baseline_df['recent_hint_deviation'] = (
    baseline_df['hint_count']
    - baseline_df['recent_hint_baseline']
)

baseline_df['recent_correct_deviation'] = (
    baseline_df['correct']
    - baseline_df['recent_accuracy_baseline']
)
```

```python
recent_cols = [
    'recent_rt_baseline',
    'recent_attempt_baseline',
    'recent_hint_baseline',
    'recent_accuracy_baseline',
    'recent_rt_deviation',
    'recent_attempt_deviation',
    'recent_hint_deviation',
    'recent_correct_deviation'
]

print(
    baseline_df[recent_cols]
    .isnull()
    .sum()
)

baseline_df[
    [
        'user_id',
        'log_response_time',
        'recent_rt_baseline',
        'recent_rt_deviation',
        'correct',
        'recent_accuracy_baseline',
        'recent_correct_deviation'
    ]
].head(30)
```

```python
recent_feature_cols = [
    'recent_rt_baseline',
    'recent_attempt_baseline',
    'recent_hint_baseline',
    'recent_accuracy_baseline',
    'recent_rt_deviation',
    'recent_attempt_deviation',
    'recent_hint_deviation',
    'recent_correct_deviation'
]

recent_df = baseline_df.dropna(
    subset=recent_feature_cols
).copy()

print("Rows after removing insufficient-history rows:", len(recent_df))
print("Confused %:", round(recent_df['confused_label'].mean() * 100, 2))
```

```python
recent_train = recent_df[
    recent_df['user_id'].isin(train_users)
].copy()

recent_val = recent_df[
    recent_df['user_id'].isin(val_users)
].copy()

recent_test = recent_df[
    recent_df['user_id'].isin(test_users)
].copy()

print("Train:", recent_train.shape)
print("Validation:", recent_val.shape)
print("Test:", recent_test.shape)

print("\nConfusion percentages:")
print("Train:", round(recent_train['confused_label'].mean() * 100, 2))
print("Validation:", round(recent_val['confused_label'].mean() * 100, 2))
print("Test:", round(recent_test['confused_label'].mean() * 100, 2))
```

```python
recent_bottom_hint_median = recent_train['bottom_hint'].median()

recent_train['bottom_hint'] = recent_train['bottom_hint'].fillna(
    recent_bottom_hint_median
)

recent_val['bottom_hint'] = recent_val['bottom_hint'].fillna(
    recent_bottom_hint_median
)

recent_test['bottom_hint'] = recent_test['bottom_hint'].fillna(
    recent_bottom_hint_median
)

print("Train missing:", recent_train.isnull().sum().sum())
print("Validation missing:", recent_val.isnull().sum().sum())
print("Test missing:", recent_test.isnull().sum().sum())
```

```python
from xgboost import XGBClassifier

final_behavior_model = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    objective='binary:logistic',
    eval_metric='aucpr',
    tree_method='hist',
    random_state=42,
    n_jobs=-1
)

final_behavior_model.fit(
    X_train,
    y_train,
    eval_set=[(X_val, y_val)],
    verbose=50
)
```

```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

val_prob_final = final_behavior_model.predict_proba(X_val)[:, 1]

threshold_results = []

for threshold in np.arange(0.50, 0.91, 0.01):

    pred = (val_prob_final >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val,
        pred
    ).ravel()

    threshold_results.append({
        "threshold": round(threshold, 2),

        "precision": precision_score(
            y_val,
            pred,
            zero_division=0
        ),

        "recall": recall_score(
            y_val,
            pred,
            zero_division=0
        ),

        "f1_confused": f1_score(
            y_val,
            pred,
            pos_label=1
        ),

        "macro_f1": f1_score(
            y_val,
            pred,
            average='macro'
        ),

        "false_positives": fp,
        "false_negatives": fn
    })

threshold_final_df = pd.DataFrame(threshold_results)

best_final = threshold_final_df.loc[
    threshold_final_df['macro_f1'].idxmax()
]

print("Best validation threshold:")
print(best_final)
```

```python
best_threshold = float(best_final['threshold'])

print("Final threshold:", best_threshold)
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score,
    f1_score
)

test_prob = final_behavior_model.predict_proba(X_test)[:, 1]

test_pred = (
    test_prob >= best_threshold
).astype(int)

print(
    "Test ROC-AUC:",
    round(
        roc_auc_score(y_test, test_prob),
        4
    )
)

print(
    "Test PR-AUC:",
    round(
        average_precision_score(
            y_test,
            test_prob
        ),
        4
    )
)

print(
    "Test Macro F1:",
    round(
        f1_score(
            y_test,
            test_pred,
            average='macro'
        ),
        4
    )
)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        test_pred,
        digits=4
    )
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        test_pred
    )
)
```
