```python
import numpy as np

# -----------------------------------------
# Helper function
# -----------------------------------------
def point_distance(p1, p2):
    return np.sqrt(
        (p1.x - p2.x) ** 2 +
        (p1.y - p2.y) ** 2
    )

# -----------------------------------------
# Eye Aspect Ratio style openness
# -----------------------------------------
def calculate_eye_openness(landmarks, eye_indices):
    p1 = landmarks[eye_indices[0]]
    p2 = landmarks[eye_indices[1]]
    p3 = landmarks[eye_indices[2]]
    p4 = landmarks[eye_indices[3]]
    p5 = landmarks[eye_indices[4]]
    p6 = landmarks[eye_indices[5]]

    # Two vertical eye distances
    vertical_1 = point_distance(p2, p6)
    vertical_2 = point_distance(p3, p5)

    # Horizontal eye width
    horizontal = point_distance(p1, p4)

    if horizontal == 0:
        return 0.0

    openness = (
        vertical_1 + vertical_2
    ) / (2.0 * horizontal)

    return openness


# MediaPipe eye landmark indices
LEFT_EYE = [33, 160, 158, 133, 153, 144]
RIGHT_EYE = [362, 385, 387, 263, 373, 380]

if result.face_landmarks:

    landmarks = result.face_landmarks[0]

    left_eye_openness = calculate_eye_openness(
        landmarks,
        LEFT_EYE
    )

    right_eye_openness = calculate_eye_openness(
        landmarks,
        RIGHT_EYE
    )

    mean_eye_openness = (
        left_eye_openness +
        right_eye_openness
    ) / 2

    print(
        "Left eye openness:",
        round(left_eye_openness, 4)
    )

    print(
        "Right eye openness:",
        round(right_eye_openness, 4)
    )

    print(
        "Mean eye openness:",
        round(mean_eye_openness, 4)
    )

else:
    print("No face landmarks available.")
```

```python
import cv2
import numpy as np

def estimate_head_pose(frame, landmarks):
    h, w = frame.shape[:2]

    # 2D facial landmark points from MediaPipe
    image_points = np.array([
        [landmarks[1].x * w,   landmarks[1].y * h],    # Nose tip
        [landmarks[152].x * w, landmarks[152].y * h],  # Chin
        [landmarks[33].x * w,  landmarks[33].y * h],   # Left eye outer corner
        [landmarks[263].x * w, landmarks[263].y * h],  # Right eye outer corner
        [landmarks[61].x * w,  landmarks[61].y * h],   # Left mouth corner
        [landmarks[291].x * w, landmarks[291].y * h],  # Right mouth corner
    ], dtype=np.float64)

    # Approximate generic 3D face model
    model_points = np.array([
        [0.0,    0.0,    0.0],      # Nose tip
        [0.0, -330.0,  -65.0],      # Chin
        [-225.0, 170.0, -135.0],    # Left eye
        [225.0, 170.0, -135.0],     # Right eye
        [-150.0, -150.0, -125.0],   # Left mouth
        [150.0, -150.0, -125.0],    # Right mouth
    ])

    focal_length = w

    camera_matrix = np.array([
        [focal_length, 0, w / 2],
        [0, focal_length, h / 2],
        [0, 0, 1]
    ], dtype=np.float64)

    distortion = np.zeros((4, 1))

    success, rotation_vector, translation_vector = cv2.solvePnP(
        model_points,
        image_points,
        camera_matrix,
        distortion,
        flags=cv2.SOLVEPNP_ITERATIVE
    )

    if not success:
        return None

    rotation_matrix, _ = cv2.Rodrigues(rotation_vector)

    angles, _, _, _, _, _ = cv2.RQDecomp3x3(rotation_matrix)

    pitch = angles[0]
    yaw   = angles[1]
    roll  = angles[2]

    return pitch, yaw, roll
```

```python
import numpy as np

def mean_point(landmarks, indices):
    xs = [landmarks[i].x for i in indices]
    ys = [landmarks[i].y for i in indices]

    return np.mean(xs), np.mean(ys)


def calculate_gaze_ratio(landmarks):

    # Iris landmarks
    LEFT_IRIS = [468, 469, 470, 471, 472]
    RIGHT_IRIS = [473, 474, 475, 476, 477]

    # Eye outer/inner corners
    LEFT_CORNERS = [33, 133]
    RIGHT_CORNERS = [362, 263]

    left_iris_x, _ = mean_point(
        landmarks,
        LEFT_IRIS
    )

    right_iris_x, _ = mean_point(
        landmarks,
        RIGHT_IRIS
    )

    left_x1 = landmarks[LEFT_CORNERS[0]].x
    left_x2 = landmarks[LEFT_CORNERS[1]].x

    right_x1 = landmarks[RIGHT_CORNERS[0]].x
    right_x2 = landmarks[RIGHT_CORNERS[1]].x

    # Normalize iris location inside each eye
    left_min = min(left_x1, left_x2)
    left_max = max(left_x1, left_x2)

    right_min = min(right_x1, right_x2)
    right_max = max(right_x1, right_x2)

    left_ratio = (
        (left_iris_x - left_min) /
        (left_max - left_min + 1e-6)
    )

    right_ratio = (
        (right_iris_x - right_min) /
        (right_max - right_min + 1e-6)
    )

    mean_ratio = (
        left_ratio + right_ratio
    ) / 2

    return left_ratio, right_ratio, mean_ratio
```

```python
import cv2
import numpy as np
import mediapipe as mp

def calculate_face_visibility(video_path, samples=30):

    cap = cv2.VideoCapture(video_path)

    total_frames = int(
        cap.get(cv2.CAP_PROP_FRAME_COUNT)
    )

    if total_frames <= 0:
        cap.release()
        return 0.0, 0, 0

    frame_indices = np.linspace(
        0,
        total_frames - 1,
        samples
    ).astype(int)

    detected = 0
    processed = 0

    options = FaceLandmarkerOptions(
        base_options=BaseOptions(
            model_asset_path="/kaggle/working/face_landmarker.task"
        ),
        running_mode=VisionRunningMode.IMAGE,
        num_faces=1,
        min_face_detection_confidence=0.5,
        min_face_presence_confidence=0.5
    )

    with FaceLandmarker.create_from_options(options) as landmarker:

        for idx in frame_indices:

            cap.set(
                cv2.CAP_PROP_POS_FRAMES,
                int(idx)
            )

            success, frame = cap.read()

            if not success:
                continue

            processed += 1

            rgb = cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2RGB
            )

            mp_image = mp.Image(
                image_format=mp.ImageFormat.SRGB,
                data=rgb
            )

            detection = landmarker.detect(mp_image)

            if detection.face_landmarks:
                detected += 1

    cap.release()

    visibility_ratio = (
        detected / processed
        if processed > 0 else 0
    )

    return visibility_ratio, detected, processed
```

```python
import cv2
import numpy as np
import mediapipe as mp


def extract_clip_features(video_path, samples=30):

    cap = cv2.VideoCapture(video_path)

    total_frames = int(
        cap.get(cv2.CAP_PROP_FRAME_COUNT)
    )

    if total_frames <= 0:
        cap.release()
        return None

    frame_indices = np.linspace(
        0,
        total_frames - 1,
        samples
    ).astype(int)

    eye_values = []
    gaze_values = []

    pitch_values = []
    yaw_values = []
    roll_values = []

    processed_frames = 0
    face_detected_frames = 0

    options = FaceLandmarkerOptions(
        base_options=BaseOptions(
            model_asset_path="/kaggle/working/face_landmarker.task"
        ),
        running_mode=VisionRunningMode.IMAGE,
        num_faces=1,
        min_face_detection_confidence=0.5,
        min_face_presence_confidence=0.5
    )

    with FaceLandmarker.create_from_options(options) as landmarker:

        for frame_index in frame_indices:

            cap.set(
                cv2.CAP_PROP_POS_FRAMES,
                int(frame_index)
            )

            success, frame = cap.read()

            if not success:
                continue

            processed_frames += 1

            rgb = cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2RGB
            )

            mp_image = mp.Image(
                image_format=mp.ImageFormat.SRGB,
                data=rgb
            )

            result_frame = landmarker.detect(mp_image)

            if not result_frame.face_landmarks:
                continue

            face_detected_frames += 1

            landmarks = result_frame.face_landmarks[0]

            # -----------------------------
            # Eye openness
            # -----------------------------

            left_eye = calculate_eye_openness(
                landmarks,
                LEFT_EYE
            )

            right_eye = calculate_eye_openness(
                landmarks,
                RIGHT_EYE
            )

            mean_eye = (
                left_eye + right_eye
            ) / 2

            eye_values.append(mean_eye)

            # -----------------------------
            # Gaze proxy
            # -----------------------------

            _, _, mean_gaze = calculate_gaze_ratio(
                landmarks
            )

            gaze_values.append(mean_gaze)

            # -----------------------------
            # Head pose
            # -----------------------------

            pose = estimate_head_pose(
                frame,
                landmarks
            )

            if pose is not None:

                pitch, yaw, roll = pose

                pitch = normalize_pitch(pitch)

                pitch_values.append(pitch)
                yaw_values.append(yaw)
                roll_values.append(roll)

    cap.release()

    if processed_frames == 0:
        return None

    face_visibility_ratio = (
        face_detected_frames /
        processed_frames
    )

    def safe_mean(values):
        arr = np.asarray(values, dtype=float)
    
        if arr.size == 0:
            return np.nan
    
        return float(np.mean(arr))


    def safe_std(values):
        arr = np.asarray(values, dtype=float)
    
        if arr.size == 0:
            return np.nan
    
        return float(np.std(arr))

    return {

        "mean_eye_openness":
            safe_mean(eye_values),

        "std_eye_openness":
            safe_std(eye_values),

        "mean_gaze_ratio":
            safe_mean(gaze_values),

        "std_gaze_ratio":
            safe_std(gaze_values),

        "mean_pitch":
            safe_mean(pitch_values),

        "mean_yaw":
            safe_mean(yaw_values),

        "mean_roll":
            safe_mean(roll_values),

        "mean_abs_pitch":
            safe_mean(np.abs(pitch_values)),

        "mean_abs_yaw":
            safe_mean(np.abs(yaw_values)),

        "mean_abs_roll":
            safe_mean(np.abs(roll_values)),

        "face_visibility_ratio":
            face_visibility_ratio,

        "processed_frames":
            processed_frames,

        "face_detected_frames":
            face_detected_frames
    }
```

```python
def extract_clip_features_fast(video_path, landmarker, samples=30):

    cap = cv2.VideoCapture(video_path)

    total_frames = int(
        cap.get(cv2.CAP_PROP_FRAME_COUNT)
    )

    if total_frames <= 0:
        cap.release()
        return None

    frame_indices = np.linspace(
        0,
        total_frames - 1,
        samples
    ).astype(int)

    eye_values = []
    gaze_values = []

    pitch_values = []
    yaw_values = []
    roll_values = []

    processed_frames = 0
    face_detected_frames = 0

    for frame_index in frame_indices:

        cap.set(
            cv2.CAP_PROP_POS_FRAMES,
            int(frame_index)
        )

        success, frame = cap.read()

        if not success:
            continue

        processed_frames += 1

        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=rgb
        )

        result_frame = landmarker.detect(mp_image)

        if not result_frame.face_landmarks:
            continue

        face_detected_frames += 1

        landmarks = result_frame.face_landmarks[0]

        # -----------------------
        # Eye openness
        # -----------------------

        left_eye = calculate_eye_openness(
            landmarks,
            LEFT_EYE
        )

        right_eye = calculate_eye_openness(
            landmarks,
            RIGHT_EYE
        )

        mean_eye = (
            left_eye + right_eye
        ) / 2

        eye_values.append(mean_eye)

        # -----------------------
        # Gaze
        # -----------------------

        _, _, mean_gaze = calculate_gaze_ratio(
            landmarks
        )

        gaze_values.append(mean_gaze)

        # -----------------------
        # Head pose
        # -----------------------

        pose = estimate_head_pose(
            frame,
            landmarks
        )
        
        if pose is not None:
        
            pitch, yaw, roll = pose
        
            pitch = normalize_pitch(pitch)
            roll = normalize_roll(roll)
        
            pitch_values.append(pitch)
            yaw_values.append(yaw)
            roll_values.append(roll)

    cap.release()

    if processed_frames == 0:
        return None

    face_visibility_ratio = (
        face_detected_frames /
        processed_frames
    )

    def safe_mean(values):

        arr = np.asarray(
            values,
            dtype=float
        )

        if arr.size == 0:
            return np.nan

        return float(
            np.mean(arr)
        )


    def safe_std(values):

        arr = np.asarray(
            values,
            dtype=float
        )

        if arr.size == 0:
            return np.nan

        return float(
            np.std(arr)
        )


    return {

        "mean_eye_openness":
            safe_mean(eye_values),

        "std_eye_openness":
            safe_std(eye_values),

        "mean_gaze_ratio":
            safe_mean(gaze_values),

        "std_gaze_ratio":
            safe_std(gaze_values),

        "mean_pitch":
            safe_mean(pitch_values),

        "mean_yaw":
            safe_mean(yaw_values),

        "mean_roll":
            safe_mean(roll_values),

        "mean_abs_pitch":
            safe_mean(np.abs(pitch_values)),

        "mean_abs_yaw":
            safe_mean(np.abs(yaw_values)),

        "mean_abs_roll":
            safe_mean(np.abs(roll_values)),

        "face_visibility_ratio":
            face_visibility_ratio,

        "processed_frames":
            processed_frames,

        "face_detected_frames":
            face_detected_frames
    }
```

```python
def process_daisee_split(
    labels_df,
    split_name,
    samples=30
):

    checkpoint_path = (
        f"/kaggle/working/"
        f"daisee_{split_name}_visual_features_checkpoint.csv"
    )

    final_path = (
        f"/kaggle/working/"
        f"daisee_{split_name}_visual_features.csv"
    )

    if os.path.exists(checkpoint_path):

        saved_df = pd.read_csv(checkpoint_path)

        rows = saved_df.to_dict(
            orient="records"
        )

        completed_ids = set(
            saved_df["ClipID"]
            .astype(str)
            .tolist()
        )

        print(
            f"{split_name}: checkpoint loaded -",
            len(completed_ids)
        )

    else:

        rows = []
        completed_ids = set()

        print(
            f"{split_name}: starting fresh"
        )

    options = FaceLandmarkerOptions(
        base_options=BaseOptions(
            model_asset_path=(
                "/kaggle/working/"
                "face_landmarker.task"
            )
        ),
        running_mode=VisionRunningMode.IMAGE,
        num_faces=1,
        min_face_detection_confidence=0.5,
        min_face_presence_confidence=0.5
    )

    start_time = time.time()

    with FaceLandmarker.create_from_options(
        options
    ) as landmarker:

        for _, row in labels_df.iterrows():

            clip_id = str(
                row["ClipID"]
            )

            if clip_id in completed_ids:
                continue

            try:

                video_path = video_lookup[
                    clip_id
                ]

                features = (
                    extract_clip_features_fast(
                        video_path,
                        landmarker,
                        samples=samples
                    )
                )

                if features is None:
                    print(
                        "FAILED -",
                        clip_id
                    )
                    continue

                features["ClipID"] = clip_id
                features["Confusion"] = row["Confusion"]

                features["confused_label"] = (
                    row["confused_label"]
                )

                rows.append(features)
                completed_ids.add(clip_id)

                completed = len(rows)

                if completed % 100 == 0:

                    pd.DataFrame(
                        rows
                    ).to_csv(
                        checkpoint_path,
                        index=False
                    )

                    mins = (
                        time.time()
                        - start_time
                    ) / 60

                    print(
                        f"{completed}/"
                        f"{len(labels_df)} "
                        f"completed | "
                        f"{mins:.1f} min"
                    )

            except Exception as e:

                print(
                    f"ERROR - {clip_id}: {e}"
                )

    result_df = pd.DataFrame(
        rows
    )

    result_df.to_csv(
        final_path,
        index=False
    )

    result_df.to_csv(
        checkpoint_path,
        index=False
    )

    print(
        f"\n{split_name.upper()} COMPLETE ✅"
    )

    print(
        "Shape:",
        result_df.shape
    )

    print(
        "Saved:",
        final_path
    )

    return result_df
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    f1_score,
    precision_score,
    recall_score,
    confusion_matrix
)

threshold_results = []

for threshold in np.arange(0.10, 0.91, 0.02):

    pred = (
        val_prob >= threshold
    ).astype(int)

    macro_f1 = f1_score(
        y_val_visual,
        pred,
        average="macro"
    )

    confused_f1 = f1_score(
        y_val_visual,
        pred,
        pos_label=1
    )

    precision = precision_score(
        y_val_visual,
        pred,
        pos_label=1,
        zero_division=0
    )

    recall = recall_score(
        y_val_visual,
        pred,
        pos_label=1,
        zero_division=0
    )

    threshold_results.append({
        "threshold": round(threshold, 2),
        "macro_f1": macro_f1,
        "confused_f1": confused_f1,
        "precision": precision,
        "recall": recall
    })

threshold_df = pd.DataFrame(
    threshold_results
)

threshold_df = threshold_df.sort_values(
    "macro_f1",
    ascending=False
)

threshold_df.head(10)
```

```python
best_threshold = threshold_df.iloc[0]["threshold"]

best_val_pred = (
    val_prob >= best_threshold
).astype(int)

print("Best threshold:", best_threshold)

print(
    "Best Macro F1:",
    round(
        f1_score(
            y_val_visual,
            best_val_pred,
            average="macro"
        ),
        4
    )
)

print(
    "Confused precision:",
    round(
        precision_score(
            y_val_visual,
            best_val_pred,
            zero_division=0
        ),
        4
    )
)

print(
    "Confused recall:",
    round(
        recall_score(
            y_val_visual,
            best_val_pred,
            zero_division=0
        ),
        4
    )
)

print(
    "Confused F1:",
    round(
        f1_score(
            y_val_visual,
            best_val_pred
        ),
        4
    )
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_val_visual,
        best_val_pred
    )
)
```
